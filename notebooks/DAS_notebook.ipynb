## 3. Order of Growth
###     1. We want to evaluate the algorithm
###     2. We want to evaluate the scalibility
###     3. We want to evaluate in terms of input size

## To calculate Order of Growth
### 1. We need to know the worst case scenario
### 2. We need to know the number of operations involved in the worst case scenario
## we use Big - O Notation  == O(n)

```python
#Example
def fact_inter(n):
    ans=1 # 1 operation
    while n>1: # 1 operation
        ans*=n # 2 operations
        n-=1 # 2 operations
    return ans

"""Now no of operation involved in fact_inter is (1+5n+1)

NOw In terms of BIG O notation it is O(n) just remove the constant parts and factors
"""
```

## Possible Order of Growth
### 1. O(1) Constant time
### 2. O(logn) Logarithmic time
### 3. O(n) Linear time
### 4. O(nlogn) Quadratic time
### 5. O(n^2) Quadratic time
### 6. O(2^n) Exponential time

### It is a collection of elements of the same data type stored at contiguous memory locations.

### The elements of an array can be accessed using a subscript operator.
### The elements of an array can be changed using the assignment operator.

## Implementing list from Scratch

```python
import ctypes
```

```python
class mera_list:
    def __init__(self):
        self.size=1   # size = no of item that can be stored
        self.n =0 # n= no. of item currently in list
        # Create a Ctype array of size = self.size
        self.A= self.__make_array(self.size)



# method to resize the array
    def __resize(self,new_capacity):
        B=self.__make_array(new_capacity)
        self.size=new_capacity
        # we have to copy all the elements from A to B
        for k in range(self.n):
            B[k]=self.A[k]
        self.A=B

    def __make_array(self,capacity):
        return (ctypes.py_object * capacity)()   # create a Ctype array of size = capacity

    # method to calculate size of list
    def __len__(self):
        return self.n

    # method to append
    def append(self,item):
        if self.n == self.size:
            self.__resize(2*self.size)
        self.A[self.n]=item
        self.n+=1


    # method to print
    def __str__(self):
        result=" "
        for i in range(self.n):
            result+=str(self.A[i])+","
        return "["+result[:-1]+"]"

    # method for indexing
    def __getitem__(self,index):
        if 0<=index<self.n:
            return self.A[index]
        else:
            raise IndexError("Index out of range")


    # method pop

    def pop(self):
        if self.n==0:
            raise IndexError("List is empty")

        print(self.A[self.n-1])
        self.n-=1


    # method to clear
    def clear(self):
        self.n=0
        self.__resize(1)

    # method to find
    def find(self,item):
        for i in range(self.n):
            if self.A[i]==item:
                return i
        raise ValueError("Item not found")

    # method to insert
    def insert(self,index,item):
        if index<0 or index>self.n:
            raise IndexError("Index out of range")
        self.__resize(2*self.size)
        for k in range(self.n,index,-1):
            self.A[k]=self.A[k-1]
        self.A[index]=item
        self.n+=1


    # method to delete
    def __delitem__(self,index):
        if index<0 or index>=self.n:
            raise IndexError("Index out of range")
        for k in range(index,self.n-1):
            self.A[k]=self.A[k+1]
        self.n-=1


    # method to remove
    def remove(self,item):
        index=self.find(item)
        if type(index)==int:
            self.__delitem__(index)
        else:
            return index

```

# Creating a linked List


```python
class node:
    def __init__(self,data):
        self.data=data
        self.next=None

class Linked_list:
    def __init__(self):
        self.head=None # condition for empty LL
        self.n=0
# Length of LL in no of nodes
    def __len__(self):  # method for length of LL
        return self.n
# insert from head
    def insert_head(self,data):
        new_node=node(data)
        new_node.next=self.head
        self.head=new_node
        self.n+=1


    def traverse(self):
        curr=self.head
        while curr !=None:
            print(curr.data)
            curr=curr.next

    def __str__(self):
        curr=self.head

        result=""
        while curr !=None:
            result=result+str(curr.data)+"->"
            curr=curr.next
        return result[:-2]

    def insert_from_tail(self,data):
        new_node=node(data)
        curr= self.head
        while curr.next!=None:
            curr=curr.next
        curr.next=new_node
        self.n+=1


    def insert_after(self,prev_node_data,data):
        new_node=node(data)

        curr=self.head
        while curr.next !=None:
            if curr.data==prev_node_data:
                new_node.next=curr.next
                curr.next=new_node
                self.n+=1
                return
            curr=curr.next

# deletion of nodes from LL

# method to clear all nodes
    def __clear__(self):
        self.head=None
        self.n=0

# method to delete after head node
    def delete_head(self):
        if self.head==None:
            raise IndexError("List is empty")
        self.head=self.head.next
        self.n-=1

# method to delete from tail node
    def delete_tail(self):
        curr=self.head
        if self.head==None:
            raise IndexError("List is empty")
        if self.n >1:
            while curr.next.next!=None:
                curr=curr.next
            curr.next = None
            self.n=self.n-1
        else:
            self.delete_head()
            self.n-=1

# method to find and remove

    def find_and_remove(self,value):

        if self.n==1:
            self.__clear__()
            return "Now List is empty"
        if self.head==None:
            return "List is empty"
        else:
            if self.head.data==value: # to handle in case value belongs to head node
                return self.delete_head()

            curr=self.head

            while curr.next !=None:
                if curr.next.data==value:
                    break
                curr=curr.next
            if curr.next==None:
                return "Not Found"
            else:
                curr.next=curr.next.next
                self.n-=1


# Searching the element in LL and return index position
    def search(self,value):
        curr=self.head
        pos=0
        while curr!=None:
            if curr.data ==value:
                return pos
            curr=curr.next
            pos+=1
        return "Not found"

# search by index position
    def __getitem__(self,index):
        curr=self.head
        pos=0
        while self.head!=None:
            if pos==index:
                return curr.data
            curr=curr.next
            pos+=1
        return "Index out of range"


    def find_the_max_and_replace_with(self,replacer_value):

        if self.head==None:
            return "List is empty"
        curr=self.head
        max_value=curr.data
        while curr!=None:
            if curr.data>max_value:
                max_value=curr.data
            curr=curr.next
        #return max_value

        curr=self.head
        while curr !=None:
            if curr.data == max_value:
                curr.data= replacer_value
            curr=curr.next

# method for reversal
    def reversal(self):
        prev_node=None
        curr_node=self.head
        while curr_node!=None:
            next_node=curr_node.next # creating a temporary pointer that points to next node
            curr_node.next=prev_node # now setting first element to behave as tail
            prev_node=curr_node
            curr_node=next_node
        self.head=prev_node


    """def statement_maker(self):
        curr_node=self.head
        #prev_node=None
        result=""
        while curr_node != None:
            result+=curr_node.data+" "
            curr_node=curr_node.next
        print(result)"""

    def statement_maker(self):
        curr_node= self.head
        prev_node=None
        result =""
        while curr_node != None:
            if curr_node.data =="*" or curr_node.data =="/":
                if prev_node.data =="*" or prev_node.data =="/":
                    pass
                result+=" "
            else:
                result+=curr_node.data
            prev_node=curr_node
            curr_node=curr_node.next

        return result

















```

# Stacks
## A simple data structure
### like LIFO - Last in First Out
# Common Operation of Stackes
###  Push
###  Pop
###  Peak
###  Size
###  isempty

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
class Stack:
    def __init__(self):
        self.top=None
        self.n=0
    # push method
    def push(self,value):
        new_node=Node(value)
        new_node.next=self.top
        self.top=new_node
        self.n+=1

    # size method
    def size(self):
        return self.n


    # isEmpty method
    def isEmpty(self):
        if self.top ==None:
            return True
        return False


    # method to traverse
    def traverse(self):
        curr=self.top
        while curr!=None:
            print(curr.data)
            curr=curr.next

    # peek pethod
    def peek(self):
        if self.isEmpty():
            return "Stack is empty"
        return self.top.data

    # pop method
    def pop(self):
        if self.isEmpty():
            return "Stack is empty"
        temp=self.top.data
        self.top=self.top.next
        self.n-=1
        return temp
```
