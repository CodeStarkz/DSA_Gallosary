## 1. Measuring Time

#### catch endtime-starttime

```python
import time

start_time=time.time()
output_holder=[]
for i in range(100):
    output_holder.append(i)
print(output_holder)
end_time=time.time()
print(f"Time taken to execute the program is {end_time-start_time}")
```

## 2. Counting Operations

### mathematical operations, Comparision operations, Assingment operations, Accessing elemnts in memory have CONSTANT time complexity


```python
# example one
# function to convert celcius to ferenhit

def c_to_f(c):
    return(c*9/5+32) # three operation involved  9/5 >> *c>> +32

# NUmber of operation in c_to_f is 3

c_to_f(100)


```

```python
# Example 2
def my_sum(n):
    total=0 # 1 operation involved
    for i in range(n+1): # 1 operation involved
        total+=i # 2 operation involved
    return total #
"""
Number of Operation involved is 1+(3n)
"""

my_sum(100)
```

## 3. Order of Growth
###     1. We want to evaluate the algorithm
###     2. We want to evaluate the scalibility
###     3. We want to evaluate in terms of input size

## To calculate Order of Growth
### 1. We need to know the worst case scenario
### 2. We need to know the number of operations involved in the worst case scenario
## we use Big - O Notation  == O(n)

```python
#Example
def fact_inter(n):
    ans=1
    while n>1: # 1 operation
        ans*=n # 2 operations
        n-=1 # 2 operations
    return ans

"""Now no of operation involved in fact_inter is (1+5n+1)

NOw In terms of BIG O notation it is O(n) just remove the constant parts and factors
"""
```

## Possible Order of Growth
### 1. O(1) Constant time
### 2. O(logn) Logarithmic time
### 3. O(n) Linear time
### 4. O(nlogn) Quadratic time
### 5. O(n^2) Quadratic time
### 6. O(2^n) Exponential time
