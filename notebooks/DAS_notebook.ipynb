## 3. Order of Growth
###     1. We want to evaluate the algorithm
###     2. We want to evaluate the scalibility
###     3. We want to evaluate in terms of input size

## To calculate Order of Growth
### 1. We need to know the worst case scenario
### 2. We need to know the number of operations involved in the worst case scenario
## we use Big - O Notation  == O(n)

```python
#Example
def fact_inter(n):
    ans=1 # 1 operation
    while n>1: # 1 operation
        ans*=n # 2 operations
        n-=1 # 2 operations
    return ans

"""Now no of operation involved in fact_inter is (1+5n+1)

NOw In terms of BIG O notation it is O(n) just remove the constant parts and factors
"""
```

## Possible Order of Growth
### 1. O(1) Constant time
### 2. O(logn) Logarithmic time
### 3. O(n) Linear time
### 4. O(nlogn) Quadratic time
### 5. O(n^2) Quadratic time
### 6. O(2^n) Exponential time

# Lets Practice some Problem and calculate their Time Complexity

```python
## Example 1

L=[i for i in range(100)]
sum=0 # 1 operation
for i in L: # n * 1 operation
    sum+=i # n * 2 operation
print(sum) # 1 operation

product =1 # 1 operation
for i in L: # n * 1 operation
    product*=i # n * 2 operation
print(product) # 1 operation


"""
No of operation of whole program is
(1+3n+1) + (1+3n+1) = 6n+4
After removing constant additive part and factors it becomes O(n).

"""
```

```python
## Example 2
L=[i for i in range(10)]
for i in L:     # n* 1 operation
    for j in L:     # n*1 operation
        print(
           ' ({},{})'.format(i,j)
        )

"""
 No of Operation
 O(n) * O(n) = O(n^2)
 so results Quadratic

"""
```

```python
## Example 3
def intToStr(i):
    digits="0123456789" # 1 operation
    if i ==0: # 1 operation
        return"0" # 1 operation
    result = ""  # 1 operation
    while i>0:  # (n* 1 +1)operation
        result = digits[i%10] + result # n*4 Operations
        i=i//10 # n*2 Operations
    return result # 1 operation

z=intToStr(1234567890)
print(z)
"""
time complexity if this function
O(logn)

expression for number of Operation
1+1+1+
7n +1 +1  where n represents no of loops executed, n= log10(i)
= 7n+5

== 7(log10(i))+5




"""
```

```python
# Example 9
def fac(n):
    if n==1:
        return 1
    else:
        return  n*fac(n-1)
print(fac(5))

"""
IN case of recursion the number of O(n) = times function is called
"""
```

# What is data Structure
## It is the way to store and organize data efficiently

# Space Efficient
## It is the way to store data in such a way that it takes less space than required


# Time Efficient
## It is the way to perform operations on data in such a way that it takes less time than required

## Linear VS Non-Linear Data Structure
### Linear (Array, Linked List,Stack,Queue )
### Non-Linear (Tree,Graph)

## Array
### Linear Data Structure, Homogeneous data
### It is a collection of elements of the same data type stored at contiguous memory locations.
### It is indexed by a non-negative integer.
### The size of an array is fixed during its creation.
### The memory allocated for an array is contiguous.
### The elements of an array can be accessed using a subscript operator.
### The elements of an array can be changed using the assignment operator.
### The time complexity of array operations is O(1).
## DisAdvantage of Array
### Fixed Size (But can be changed via Dynamic Array )
### Homogeneous data (All elements are of same data type). But can be heterogeneous via Referential Array.
## List in python is nothing but an an array (with the implementation of Dynamic & Referential Array )
